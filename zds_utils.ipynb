```python
# Cortar un vídeo en en fragmentos de x segundos
from moviepy.video.io.VideoFileClip import VideoFileClip
video_path = "inputs/T01E08/ori/T01E08.mp4"
from moviepy.video.io.ffmpeg_tools import ffmpeg_extract_subclip
def split_video_into_chunks(video_path, chunk_duration=180, start=0, end=None):
		"""Splits a video into chunks of specified duration."""
		video = VideoFileClip(video_path)
		duration = int(video.duration)
		
		for s in range(start, duration, chunk_duration):
				e = min(s + chunk_duration, duration)
				if end is not None and e > end:
					break
				#output_path = f"inputs/T01E07/split/T01E07_{s // 60:02d}_{e // 60:02d}.mp4"
				output_path = f"inputs/T01E08/split/T01E08_{e // 90:02d}.mp4"
				ffmpeg_extract_subclip(video_path, s, e, targetname=output_path)
				print(f"Chunk saved: {output_path}")
split_video_into_chunks(video_path, chunk_duration=180, start=55)
```

```python
#@title Extract frames
import cv2
import os
from urllib.request import urlretrieve

def extract_frames(video_url, output_path=None, frames_per_second=10, width=None, height=None, extension="jpg"):
    """
    Extracts frames from a video URL at a specified frame rate, optionally resizes them,
    and saves them to a folder named after the video file in the output path.

    Args:
        video_url (str): The URL of the input video file.
        output_path (str): The directory to save the extracted frames.
        frames_per_second (int, optional): The number of frames to extract per second. Defaults to 10.
        width (int, optional): The desired width to resize the frames to. Defaults to None.
        height (int, optional): The desired height to resize the frames to. Defaults to None.
        extension (str optiona): Type of file generated. Default to jpg.
    """
    try:
        # Download the video from the URL
        video_name = video_url.split('/')[-1]

        # Determine the output path
        if output_path is None:
            output_path = os.path.dirname(video_url)

        # Create output directory
        video_folder_name = os.path.splitext(video_name)[0]
        output_folder = os.path.join(output_path, video_folder_name)
        if not os.path.exists(output_folder):
            os.makedirs(output_folder)

        # Open the video file
        cap = cv2.VideoCapture(video_url)
        if not cap.isOpened():
            print(f"Error: Could not open video file {video_url}")
            return

        # Get video properties
        fps = cap.get(cv2.CAP_PROP_FPS)
        frame_count = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        duration = frame_count / fps

        print(f"Video properties: FPS={fps}, Frame Count={frame_count}, Duration={duration:.2f} seconds")

        # Calculate frame interval for extraction
        frame_interval = max(1, int(fps / frames_per_second))

        frame_number = 0
        extracted_frame_count = 0

        while True:
            ret, frame = cap.read()
            if not ret:
                break

            # Extract frame at specified interval
            if frame_number % frame_interval == 0:
                # Resize frame if width and height are provided
                if width is not None and height is not None:
                    frame = cv2.resize(frame, (width, height))

                output_filename = os.path.join(output_folder, f"frame_{extracted_frame_count:06d}.{extension}")
                cv2.imwrite(output_filename, frame)
                extracted_frame_count += 1

            frame_number += 1

        cap.release()
        print(f"Frame extraction complete. Extracted {extracted_frame_count} frames to {output_folder}")

    except Exception as e:
        print(f"An error occurred: {e}")
```

```python
# Unir vídeos
from moviepy.editor import VideoFileClip, concatenate_videoclips
def merge_videos(video_paths, output_path):
		"""
		Merges multiple video files into a single video file.

		Args:
				video_paths (list): List of paths to the video files to be merged.
				output_path (str): Path where the merged video will be saved.
		"""
		clips = [VideoFileClip(video) for video in video_paths]
		final_clip = concatenate_videoclips(clips, method="compose")
		final_clip.write_videofile(output_path, codec="libx264", audio_codec="aac")
		print(f"Merged video saved at {output_path}")

# Ejemplo de uso
video_paths = ["F:\IA\ZDS\inputs\T01E07\Intro01.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro02.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro03.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro04.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro05.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro06.mp4",
							"F:\IA\ZDS\inputs\T01E07\Intro07.mp4"]
output_path = "F:\IA\ZDS\inputs\T01E07\MergedIntro.mp4"
merge_videos(video_paths, output_path)
```

```python
from moviepy.editor import VideoFileClip, concatenate_videoclips
import os

def repeat_video(input_path, repeat_count):
	"""
	Repeats a video multiple times, including fractional repeats.

	Args:
		input_path (str): Path to the input video file.
		repeat_count (float): Number of times to repeat the video (can be fractional).

	Returns:
		str: Path to the output repeated video.
	"""
	video = VideoFileClip(input_path)
	full_repeats = int(repeat_count)
	partial = repeat_count - full_repeats

	clips = [video] * full_repeats

	if partial > 0:
		partial_duration = video.duration * partial
		clips.append(video.subclip(0, partial_duration))

	final_clip = concatenate_videoclips(clips)
	base, ext = os.path.splitext(input_path)
	output_path = f"{base}_repeated{repeat_count:.2f}{ext}"
	final_clip.write_videofile(output_path, codec="libx264", audio_codec="aac")
	print(f"✅ Video repetido guardado en {output_path}")
	return output_path

# Ejemplo de uso:
video_path = "inputs/T01E08/T01E08_Cuerpo.mp4"
repeat_video(video_path, 30)
```

```python
# Agregar texto a un vídeo
# Permite agregar un texto en una posición específica del vídeo, con un tamaño y color determinados.
from moviepy.editor import VideoFileClip, TextClip, CompositeVideoClip
def add_text_to_video(video_path, text, position=("center", "bottom"), font_size=24, color="white", output_path=None):
		"""
		Adds text to a video at a specified position.

		Args:
				video_path (str): Path to the input video file.
				text (str): Text to be added to the video.
				position (tuple): Position of the text in the format (x, y).
				font_size (int): Size of the text font.
				color (str): Color of the text.
				output_path (str): Path to save the output video with text. If None, saves in the same directory as input.
		"""
		# Load the video
		video = VideoFileClip(video_path)

		# Create a TextClip
		txt_clip = TextClip(text, fontsize=font_size, color=color)
		txt_clip = txt_clip.set_position(position).set_duration(video.duration)

		# Overlay the text on the video
		video_with_text = CompositeVideoClip([video, txt_clip])

		# Set output path if not provided
		if output_path is None:
				output_path = f"{os.path.splitext(video_path)[0]}_with_text.mp4"

		# Write the result to a file
		video_with_text.write_videofile(output_path, codec="libx264", audio_codec="aac")
		print(f"Video with text saved at {output_path}")
# Ejemplo de uso
video_path = "F:\IA\ZDS\inputs\T01E07\MergedIntro.mp4"
text = "Singularidad, Día Cero."
position = ("center", "bottom")
font_size = 36
color = "white"
output_path = "F:\IA\ZDS\inputs\T01E07\MergedIntro_with_text.mp4"
add_text_to_video(video_path, text, position, font_size, color, output_path)
```
